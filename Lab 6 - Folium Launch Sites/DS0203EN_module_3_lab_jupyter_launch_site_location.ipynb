```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Define geographical coordinates
railway_lat = 28.56529
railway_lon = -80.58654

highway_lat = 28.56371
highway_lon = -80.57084

city_lat = 28.54228
city_lon = -81.38055

# Define geographical features
proximities = [
    ('Railway', railway_lat, railway_lon, 'purple'),
    ('Highway', highway_lat, highway_lon, 'blue'),
    ('City', city_lat, city_lon, 'orange')
]

for name, lat, lon, color in proximities:

    # Calculate distance
    distance = calculate_distance(
        launch_site_lat, launch_site_lon, lat, lon
    )

    # Create distance marker
    distance_marker = folium.Marker(
        location=[lat, lon],
        icon=DivIcon(
            icon_size=(150, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size:12px; color:#d35400;"><b>%s</b></div>'
            % "{}: {:.2f} KM".format(name, distance)
        )
    )

    site_map.add_child(distance_marker)

    # Draw distance line
    lines = folium.PolyLine(
        locations=[
            [launch_site_lat, launch_site_lon],
            [lat, lon]
        ],
        weight=2,
        color=color
    )

    site_map.add_child(lines)

    print("{}: {:.2f} KM".format(name, distance))

# Display the map
site_map
```

The interactive geographical analysis using Folium suggests that transportation accessibility, coastal proximity and public safety are important considerations in rocket launch-site selection.

Railway accessibility can facilitate the transportation of heavy equipment and rocket components.

Highway accessibility supports efficient transportation of personnel, materials and supplies.
